## Pure Python for science

```python
from math import sin

def my_func(x):
    return sin(x) + x**2 + 5 
```

```python
my_func(3)
```

```python
x_values = [1, 2, 3, 4] # CREATING A LIST : LONG AND PAINFUL
x_values
```

```python
my_func(x_values)
```

```python
y_values = []
for xv in x_values:
    y = my_func(xv)
    y_values.append(y)
y_values
```

### Use of Numpy

```python
import numpy as np
```

```python
def my_func2(x):
    return np.sin(x) + x**2 + 5 
```

```python
x_array = np.array(x_values)
x_array
```

```python
my_func2(x_array)
```

## Plotting with Matplotlib

```python
%matplotlib widget
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0., 10., 100) # X VALUES
y = my_func2(x)
```

```python
fig = plt.figure()
plt.plot(x, y)
plt.grid() 
plt.xlabel("Time, $x$")
plt.ylabel("Signal amplitude, $y$")
plt.title("Awesome data")
plt.show()
```
